```python
from zemberek import TurkishMorphology, TurkishTokenizer

tokenizer = TurkishTokenizer.DEFAULT
morphology = TurkishMorphology.create_with_defaults()

text = "Kitapları okuyamadığımız için öğretmenimiz üzüldü."

# Tokenization
tokens = [t.content for t in tokenizer.tokenize(text)]
print(tokens)

# Bağlama göre en olası çözümleme + lemma
analysis = morphology.analyze_and_disambiguate(text).best_analysis()
for a in analysis:
    print(a.item.lemma, "->", a.format_string())
```
