```python
def clean_text(text):
    output = text.str.replace(r'https?://\S+|www\.\S+', '', regex=True)    
    output=output.str.lower()
    output=output.str.replace('\xa0', ' ')
    output=output.str.replace(r'[^\w\s]', '', regex=True)
    return output
```

```python
# Karakter uzunluğu
df['char_length'] = df['clean'].astype(str).apply(len)

# Kelime sayısı
df['word_count'] = df['clean'].astype(str).apply(lambda x: len(x.split()))

# Özet istatistikler (ortalama, medyan, min, max vb.)
print(df[['char_length', 'word_count']].describe())
```
